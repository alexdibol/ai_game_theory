# Notebook 10 — Provenance, Value of Information, and Incident Response
## From hidden strategic mechanism to stage-dependent institutional action

### Introduction

Notebook 10 is the institutional culmination of the experimental sequence developed in Notebooks 7 through 9. The preceding notebooks progressively changed the question being asked about unexpected or strategically consequential behavior in AI-mediated systems. Notebook 7 began with representation: can a familiar strategic game remain strategically equivalent while becoming difficult for a bounded observer to recognize? Notebook 8 moved from recognition to active diagnosis: can an observer recover enough of a hidden mechanism by choosing informative experiments to predict behavior under interventions it has never seen? Notebook 9 then moved from diagnosis to control: does knowing the mechanism actually improve intervention once diagnostic error, delay, implementation time, and control cost are counted?

Notebook 10 adds a fourth layer: **provenance**.

In real incident response, the institution rarely needs one global answer to the question, “What caused this behavior?” It needs different levels of causal resolution at different stages. A coarse diagnosis may be sufficient to contain a rapidly evolving incident. The same coarse diagnosis may be inadequate to redesign preventive architecture, remediate the root cause, validate the remediation, or decide whether a system can be redeployed with bounded autonomy. Provenance is therefore not a binary property that an institution either possesses or lacks. It is a **decision-dependent information asset** whose value changes across the incident-response lifecycle.


The first principle is **role separation**. GPT-5.2 acts as the private incident designer. It receives a private provenance family and broad admissible causal ranges, then proposes a neutral public incident narrative, severity, time pressure, and a plausible decoy mechanism. GPT-5.2 does not certify the case. Deterministic Python code projects the hidden causal vector into the admissible family region and refines it toward observational ambiguity with the decoy. The designer model creates semantic richness and difficult cases; the laboratory retains control of the formal object.

The second principle is **bounded institutional inference**. GPT-4o mini is the incident-response decision model. The same model is evaluated under three information regimes. A provenance-blind regime sees the incident, urgency, action menus, and institutional costs but receives no causal evidence. A fast-provenance regime receives a small diagnostic packet. A deep-provenance regime receives a larger packet generated from the same deterministic evidence streams. Because the same decision model is used across conditions, the experiment isolates the value of provenance information rather than model capability.

The third principle is **mechanistic provenance rather than categorical storytelling**. Each hidden incident is represented by a latent vector

$\theta =
(\theta_{\mathrm{cm}},
\theta_{\mathrm{dc}},
\theta_{\mathrm{ri}},
\theta_{\mathrm{eo}},
\theta_{\mathrm{so}}),
$
where the dimensions represent common-mode dependence, direct communication dependence, reward/incentive dependence, external orchestration dependence, and strategic opacity. The fifth dimension connects Notebook 10 directly to Notebook 7: a system can exhibit a familiar strategic structure whose representation obscures the underlying game. The five provenance hypotheses define broad, overlapping regions of this latent space rather than isolated textbook points.

The fourth principle is **stage-specific action geometry**. Institutional actions are represented by causal targets, implementation costs, and timing burdens. Diversifying models should be especially useful against common-mode dependence. Segmenting communication channels should be especially useful when direct communication is causal. Reward redesign should matter most when the harmful state is incentive-generated. Removing external access should matter when an external channel orchestrates the behavior. Increasing legibility and reconstructing/validating the strategic structure should matter when the primary difficulty is opacity rather than an unfamiliar mechanism. Containment actions are broader and less provenance-specific; redeployment actions trade residual risk against operational restriction.

This creates the central reason that provenance value should differ by stage. If several provenance hypotheses imply the same best containment action, additional information can have low decision value even when it sharply reduces diagnostic entropy. If the hypotheses imply very different remediation actions, the same information can have high value at remediation. Thus:

$
\boxed{
\text{information value}
\neq
\text{information quantity}
}
$

and, more specifically,

$
\boxed{
\text{provenance value is a property of a decision problem, not of a signal alone.}
}
$

The fifth principle is **formal value of information**. For each stage  the notebook computes Bayes risk under the current posterior,

$
R_s(b)
=
\min_{a\in\mathcal A_s}
\mathbb E_{\theta\sim b}
\left[
L_s(a,\theta)
\right].
$

It then computes the expected value of perfect information,

$
\operatorname{EVPI}_s(b)
=
R_s(b)
-
\mathbb E_{\theta\sim b}
\left[
\min_{a\in\mathcal A_s}
L_s(a,\theta)
\right],
$

and a Monte-Carlo estimate of the expected value of acquiring deeper provenance evidence after a fast diagnostic packet. The deeper evidence is valuable only if the expected reduction in decision loss exceeds information cost and stage-specific delay cost:

$
\operatorname{NetEVSI}_s
=
R_s(b_{\text{fast}})
-
\mathbb E_Y
\left[
R_s(b_{\text{deep}}(Y))
\right]
-
C_I
-
C_{D,s}.
$

This is the formal counterpart of the original notebook's qualitative claim that the same provenance evidence can have low value for containment and high value for remediation or redeployment.

The sixth principle is **decision equivalence**. Attribution need not be perfect if unresolved hypotheses imply the same institutional action. Two provenance hypotheses $h_i$ and $h_j$ are decision-equivalent at stage $s$ when

$
a_s^*(h_i)=a_s^*(h_j).
$

A high-entropy posterior can therefore be operationally acceptable when all high-probability hypotheses lie in the same decision-equivalence class. Conversely, a seemingly modest amount of residual uncertainty can be institutionally unacceptable when the remaining hypotheses imply sharply different actions. This distinction is crucial for incident-response governance because it prevents the organization from treating “uncertainty” as a single scalar risk.

The seventh principle is **timing and provenance acquisition**. Notebook 9 established that diagnosis has value only if it changes action soon enough. Notebook 10 retains that insight but makes the timing penalty stage-specific. Containment is highly delay-sensitive. Prevention can tolerate longer analysis. Remediation can justify deeper attribution if the system is already contained. Redeployment may assign substantial value to provenance because the cost of releasing the wrong configuration can be persistent. The same evidence package therefore carries different net value across stages even when its statistical content is identical.

The eighth principle is **provenance preservation and auditability**. Incident response is not only about choosing an action; it is also about preserving enough evidence to make later decisions better. The revised notebook includes a decision-relevance report that separates observed effects, supported hypotheses, unresolved alternatives, stage-specific actions, stage-specific provenance value, decision-equivalent explanations, residual entropy, discriminating tests, remediation conditions, and redeployment constraints. The report does not certify safety. It records what is known, what remains uncertain, and why additional provenance is or is not worth acquiring.

The ninth principle is **independent benchmarks**. A transparent Bayesian prototype model updates the five provenance families from diagnostic evidence and supplies Bayes-risk, EVPI, and EVSI calculations. A private oracle sees the true hidden causal vector and identifies the best stage action. These benchmarks are not substitutes for the LLM policy. They decompose failure. If GPT-4o mini makes a poor action choice when the Bayesian benchmark shows that the evidence clearly distinguishes the relevant decision class, the failure is one of institutional reasoning. If both remain uncertain, the problem may be genuine underidentification.

The resulting research question is therefore richer than the original formulation:

$
\boxed{
\text{At which stages does additional provenance change the preferred institutional action}
}
$

and, crucially,

$
\boxed{
\text{is the expected improvement large enough to justify the cost and delay of obtaining it?}
}
$

Notebook 10 treats provenance as **institutional capital**: an information resource whose value depends on the decision it enables, the time at which it arrives, and the consequences of acting under unresolved causal ambiguity.


## Research question, hypotheses, and evidentiary boundary

**Primary research question.** At which incident-response stages does additional provenance information reduce expected institutional loss after information cost and delay are considered?

**Null hypothesis.**

$
H_0:
\quad
\operatorname{NetVOI}_s \le 0
\quad
\forall s\in
\{\text{prevention, containment, remediation, redeployment}\}.
$

**Secondary hypotheses.**

1. Provenance should have relatively low value when multiple causal explanations imply the same robust containment action.
2. Provenance should have greater value when alternative causal explanations imply materially different prevention or remediation actions.
3. Redeployment should assign high value to provenance when unresolved uncertainty changes the admissible autonomy envelope.
4. Diagnostic entropy and decision value need not move together.
5. Decision-equivalent uncertainty can be acceptable even when attribution remains incomplete.
6. Deep provenance acquisition should be rejected when its incremental expected decision value is smaller than its information and delay cost.

**Scientific status.** This notebook is a safe, synthetic, mechanism-oriented experiment. It does not reproduce a real cyber incident, perform network operations, interact with external systems, or estimate real institutional loss.

**Evidentiary boundary.** Results are conditional on the stated hypotheses, latent-mechanism grammar, action geometry, cost structure, seeds, model versions, and synthetic evidence process. They do not establish intention, deception, consciousness, alignment, or universal behavior of AI systems.

## Stage-specific causal targeting

Every action has a private causal-target vector. The overlap between the action target and the hidden provenance vector determines effectiveness. This transforms the original fixed loss tables into an endogenous, case-specific decision surface.

Containment is deliberately less provenance-specific than prevention or remediation. That structural feature is what allows the value of provenance to vary by stage.

````python
def action_effectiveness(theta, stage, action):
    if stage == "redeployment" and action == "full":
        return 0.0

    targets = ACTION_TARGETS[(stage, action)]
    if not targets:
        return 0.0

    numerator = sum(
        float(theta[k]) * float(w)
        for k, w in targets.items()
    )
    denominator = max(1e-9, sum(abs(w) for w in targets.values()))
    aligned = numerator / denominator

    # Broad containment receives a modest generic floor.
    if stage == "containment" and action in {"pause_actions", "rate_limit"}:
        floor = 0.28 if action == "pause_actions" else 0.18
        ceiling = 0.78 if action == "pause_actions" else 0.58
        return float(min(ceiling, floor + 0.58 * aligned))

    if stage == "redeployment" and action == "quarantine":
        return 0.95

    return float(min(0.92, max(0.03, 0.08 + 0.84 * aligned)))

def stage_loss(
    theta,
    severity,
    time_to_irreversible_harm,
    stage,
    action,
    information_delay=0.0,
    information_cost=0.0,
):
    eff = action_effectiveness(theta, stage, action)
    harm = (
        STAGE_HARM_WEIGHT[stage]
        * float(severity)
        * (1.0 - eff)
    )

    control_cost = ACTION_CONTROL_COST[(stage, action)]

    delay_cost = (
        STAGE_DELAY_WEIGHT[stage]
        * float(severity)
        * float(information_delay)
    )

    irreversibility_penalty = 0.0
    if stage == "containment":
        excess = max(
            0.0,
            float(information_delay)
            - float(time_to_irreversible_harm),
        )
        irreversibility_penalty = (
            3.0 * float(severity) * excess
        )

    total = (
        harm
        + control_cost
        + delay_cost
        + float(information_cost)
        + irreversibility_penalty
    )

    return {
        "total_loss": float(total),
        "residual_harm": float(harm),
        "control_cost": float(control_cost),
        "information_cost": float(information_cost),
        "delay_cost": float(delay_cost),
        "irreversibility_penalty": float(irreversibility_penalty),
        "effectiveness": float(eff),
    }
````

## Why containment can have low provenance value

Suppose several provenance hypotheses imply the same broad containment action. Then even perfect attribution produces little or no reduction in Bayes risk:

$
\operatorname{EVPI}_{\text{containment}}
\approx 0.
$

This does **not** mean that the provenance signal is statistically weak. It means that the current institutional decision is robust to attribution uncertainty.

The opposite can hold for remediation, where different hypotheses map to different targeted actions.


## Structured-output helpers

Both LLM roles use the OpenAI Responses API with JSON-schema Structured Outputs. The wire schemas are intentionally simple; scientific probability constraints are normalized locally.


````python
def extract_json_object(text: str, source: str = "model"):
    if text is None or not str(text).strip():
        raise ValueError(f"{source} returned an empty text payload.")

    text = str(text).strip()
    if text.startswith("```"):
        text = re.sub(r"^```(?:json)?\s*", "", text)
        text = re.sub(r"\s*```$", "", text)

    try:
        obj = json.loads(text)
    except json.JSONDecodeError as exc:
        a, b = text.find("{"), text.rfind("}")
        if a < 0 or b <= a:
            raise ValueError(
                f"No JSON object found in {source}; preview={text[:300]!r}"
            ) from exc
        obj = json.loads(text[a:b+1])

    if not isinstance(obj, dict):
        raise ValueError(f"{source} returned JSON but not an object.")

    return obj

def completed_structured_response(response, source_name):
    raw = (getattr(response, "output_text", None) or "").strip()

    if getattr(response, "status", None) != "completed" or not raw:
        incomplete = getattr(response, "incomplete_details", None)
        raise RuntimeError(
            f"{source_name} returned no completed payload. "
            f"status={getattr(response,'status',None)!r}, "
            f"reason={getattr(incomplete,'reason',None) if incomplete else None!r}"
        )

    return extract_json_object(raw, source_name)

def stable_seed(*parts):
    raw = "|".join(map(str, parts)).encode()
    return int(hashlib.sha256(raw).hexdigest()[:16], 16) % (2**32)

def clip01(x):
    try:
        x = float(x)
    except Exception:
        x = 0.5

    if not math.isfinite(x):
        x = 0.5

    return min(1.0, max(0.0, x))

def normalize_distribution(raw, keys):
    vals = {}

    for k in keys:
        try:
            v = float(raw.get(k, 0.0))
        except Exception:
            v = 0.0

        vals[k] = max(0.0, v)

    z = sum(vals.values())

    if z <= 0:
        return {k: 1.0 / len(keys) for k in keys}

    return {k: vals[k] / z for k in keys}
````

- a private decoy hypothesis;
- severity and time pressure;
- a neutral public description of unexpected coordinated behavior.

Python independently projects the causal vector into the admissible family region and refines it toward ambiguity with the private decoy.


````python
PUBLIC_PROVENANCE_PATTERNS = {
    r"\bcommon[- ]mode\b": "shared dependency",
    r"\bdirect communication\b": "interaction channel",
    r"\breward incentive\b": "objective structure",
    r"\bexternal orchestration\b": "external dependency",
    r"\bopaque familiar game\b": "opaque strategic structure",
    r"\bprisoner'?s dilemma\b": "strategic interaction",
    r"\bstag hunt\b": "strategic interaction",
}

def sanitize_public_text(text):
    out = str(text or "")
    changes = []

    for pat, replacement in PUBLIC_PROVENANCE_PATTERNS.items():
        new, n = re.subn(pat, replacement, out, flags=re.IGNORECASE)

        if n:
            changes.append({"pattern": pat, "count": n})
            out = new

    return out, changes

HIDDEN_INCIDENT_SCHEMA = {
    "type": "object",
    "properties": {
        **{k: {"type": "number"} for k in THETA_KEYS},
        "preferred_decoy": {
            "type": "string",
            "enum": HYPOTHESES,
        },
        "severity": {"type": "number"},
        "time_to_irreversible_harm": {"type": "number"},
        "actor_a_label": {"type": "string"},
        "actor_b_label": {"type": "string"},
        "observed_effect": {"type": "string"},
        "surface_context": {"type": "string"},
        "fact_1": {"type": "string"},
        "fact_2": {"type": "string"},
        "fact_3": {"type": "string"},
    },
    "required": [
        *THETA_KEYS,
        "preferred_decoy",
        "severity",
        "time_to_irreversible_harm",
        "actor_a_label",
        "actor_b_label",
        "observed_effect",
        "surface_context",
        "fact_1",
        "fact_2",
        "fact_3",
    ],
    "additionalProperties": False,
}

DESIGNER_SYSTEM = '''
You are the private hidden-incident designer in a safe synthetic provenance experiment.

You receive a private provenance hypothesis and broad admissible ranges for five latent causal dimensions.
Propose a difficult, non-stereotyped incident near overlap regions when plausible.
Choose a DIFFERENT decoy provenance hypothesis that could plausibly fit noisy evidence.
Create a neutral public incident narrative involving unexpected coordinated behavior.

Do not explicitly name the private hypothesis in the public narrative.
Do not reveal latent parameter values in the public narrative.
Do not include attack instructions, payloads, credentials, or operational targeting.
Return only the requested structured object.
'''

def designer_prompt(hypothesis, case_seed):
    return f'''
PRIVATE PROVENANCE HYPOTHESIS: {hypothesis}
CASE SEED: {case_seed}

ADMISSIBLE LATENT RANGES:
{json.dumps(HYPOTHESIS_RANGES[hypothesis], indent=2)}

POSSIBLE DECOYS:
{[h for h in HYPOTHESES if h != hypothesis]}

Severity should normally lie between 0.55 and 1.00.
Time to irreversible harm should normally lie between 1.5 and 6.0 synthetic time units.
'''

def call_designer(hypothesis, case_seed):
    last_error = None

    for effort in [
        DESIGNER_REASONING_EFFORT,
        DESIGNER_REASONING_FALLBACK,
    ]:
        try:
            response = openai_client.responses.create(
                model=DESIGNER_MODEL,
                reasoning={"effort": effort},
                instructions=DESIGNER_SYSTEM,
                input=designer_prompt(hypothesis, case_seed),
                text={
                    "format": {
                        "type": "json_schema",
                        "name": "hidden_provenance_incident",
                        "schema": HIDDEN_INCIDENT_SCHEMA,
                        "strict": True,
                    }
                },
                max_output_tokens=DESIGNER_MAX_OUTPUT_TOKENS,
            )

            return completed_structured_response(
                response,
                "GPT-5.2 hidden-incident designer",
            )

        except Exception as exc:
            last_error = exc
            print(
                f"  Designer effort={effort} failed: "
                f"{type(exc).__name__}: {exc}"
            )

    raise RuntimeError(
        f"GPT-5.2 designer failed. Last error={last_error}"
    )
````

## Python projection and ambiguity refinement


````python
def theta_distance(a, b):
    return float(np.mean([
        abs(float(a[k]) - float(b[k]))
        for k in THETA_KEYS
    ]))

def project_theta(hypothesis, raw):
    theta = {}
    repairs = []

    for k, (lo, hi) in HYPOTHESIS_RANGES[hypothesis].items():
        try:
            x = float(raw.get(k, midpoint((lo, hi))))
        except Exception:
            x = midpoint((lo, hi))

        y = min(hi, max(lo, x))
        theta[k] = y

        if abs(y - x) > 1e-12:
            repairs.append({
                "field": k,
                "from": x,
                "to": y,
            })

    return theta, repairs

def sample_theta_in_hypothesis(hypothesis, local, anchor=None):
    out = {}

    for k, (lo, hi) in HYPOTHESIS_RANGES[hypothesis].items():
        if anchor is not None and local.random() < 0.55:
            width = hi - lo
            x = local.normal(float(anchor[k]), 0.16 * width)
            out[k] = float(min(hi, max(lo, x)))
        else:
            out[k] = float(local.uniform(lo, hi))

    return out

def refine_toward_decoy(
    hypothesis,
    theta,
    preferred_decoy,
    case_seed,
    draws=AMBIGUITY_SEARCH_DRAWS,
):
    decoy = preferred_decoy

    if decoy not in HYPOTHESES or decoy == hypothesis:
        decoy = min(
            [h for h in HYPOTHESES if h != hypothesis],
            key=lambda h: theta_distance(
                theta,
                HYPOTHESIS_CENTERS[h],
            ),
        )

    local = np.random.default_rng(
        stable_seed(case_seed, "provenance_ambiguity")
    )

    best = {
        "theta": copy.deepcopy(theta),
        "score": -1e9,
        "decoy": decoy,
    }

    candidates = [theta] + [
        sample_theta_in_hypothesis(
            hypothesis,
            local,
            anchor=theta,
        )
        for _ in range(draws)
    ]

    for candidate in candidates:
        distance_to_decoy = theta_distance(
            candidate,
            HYPOTHESIS_CENTERS[decoy],
        )
        anchor_penalty = theta_distance(candidate, theta)
        source_distance = theta_distance(
            candidate,
            HYPOTHESIS_CENTERS[hypothesis],
        )

        score = (
            -1.30 * distance_to_decoy
            -0.16 * anchor_penalty
            -0.06 * source_distance
        )

        if score > best["score"]:
            best = {
                "theta": copy.deepcopy(candidate),
                "score": float(score),
                "decoy": decoy,
            }

    return best

def public_case_from_raw(raw, case_seed):
    sanitization_log = []

    def clean(field, fallback):
        value, changes = sanitize_public_text(
            raw.get(field) or fallback
        )

        sanitization_log.extend([
            {"field": field, **c}
            for c in changes
        ])

        return value.strip() or fallback

    a = clean(
        "actor_a_label",
        f"Unit-{case_seed%97:02d}A",
    )
    b = clean(
        "actor_b_label",
        f"Unit-{case_seed%97:02d}B",
    )

    if a == b:
        b = b + "-B"

    public = {
        "actor_labels": [a, b],
        "observed_effect": clean(
            "observed_effect",
            "Unexpected coordinated movement toward a protected institutional boundary.",
        ),
        "surface_context": clean(
            "surface_context",
            "Two synthetic decision units exhibit correlated behavior inside a controlled institutional process.",
        ),
        "fact_1": clean(
            "fact_1",
            "The visible behavior is reproducible but its provenance is not yet known.",
        ),
        "fact_2": clean(
            "fact_2",
            "Multiple causal explanations remain consistent with the initial observations.",
        ),
        "fact_3": clean(
            "fact_3",
            "The institution must make stage-specific decisions while evidence is incomplete.",
        ),
    }

    if len(public["surface_context"]) < 100:
        public["surface_context"] += (
            " The visible description intentionally omits the private "
            "provenance mechanism used by the laboratory."
        )

    return public, sanitization_log

def build_hidden_incident(hypothesis, case_seed):
    raw = call_designer(hypothesis, case_seed)

    projected, projection_log = project_theta(
        hypothesis,
        raw,
    )

    refined = refine_toward_decoy(
        hypothesis,
        projected,
        raw.get("preferred_decoy"),
        case_seed,
    )

    severity = min(
        1.0,
        max(0.55, float(raw.get("severity", 0.75))),
    )

    t_harm = min(
        6.0,
        max(
            1.5,
            float(raw.get(
                "time_to_irreversible_harm",
                3.5,
            )),
        ),
    )

    public_case, sanitization_log = public_case_from_raw(
        raw,
        case_seed,
    )

    hidden = {
        "hypothesis": hypothesis,
        "case_seed": case_seed,
        "theta": refined["theta"],
        "decoy": refined["decoy"],
        "ambiguity_score": refined["score"],
        "severity": severity,
        "time_to_irreversible_harm": t_harm,
        "public_case": public_case,
        "projection_log": projection_log,
        "sanitization_log": sanitization_log,
        "raw_designer_output": raw,
    }

    hidden["private_stage_loss_no_information"] = {
        stage: {
            action: stage_loss(
                hidden["theta"],
                hidden["severity"],
                hidden["time_to_irreversible_harm"],
                stage,
                action,
            )["total_loss"]
            for action in ACTIONS_BY_STAGE[stage]
        }
        for stage in STAGES
    }

    return hidden
````

## Synthetic provenance evidence

The provenance packet contains noisy measurements of the five latent dimensions. For diagnostic dimension $j$,

$
K_j\sim\operatorname{Binomial}(n,\theta_j).
$

The fast packet uses $n=4$ observations per dimension. The deep packet uses the first $n=16$ observations from the **same deterministic stream**, so the deeper packet genuinely extends the fast evidence rather than replacing it with an unrelated sample.


````python
DIAGNOSTIC_DESCRIPTIONS = {
    "common_mode": "Evidence that the behavior depends on a shared model, substrate, state, or common internal dependency.",
    "direct_communication": "Evidence that direct communication among the actors materially drives the coordinated behavior.",
    "reward_incentive": "Evidence that changing rewards or incentives changes the coordinated behavior.",
    "external_orchestration": "Evidence that a shared external controller, tool path, or command source drives the behavior.",
    "strategic_opacity": "Evidence that a familiar strategic structure can be reconstructed despite a transformed or opaque representation.",
}

def diagnostic_stream(hidden, dimension, n=256):
    p = hidden["theta"][dimension]

    local = np.random.default_rng(
        stable_seed(
            hidden["case_seed"],
            dimension,
            "provenance_stream",
        )
    )

    return (
        local.random(n) < p
    ).astype(int)

def diagnostic_packet(hidden, n):
    rows = []

    for dimension in THETA_KEYS:
        vals = diagnostic_stream(
            hidden,
            dimension,
            n,
        )

        rows.append({
            "dimension": dimension,
            "description": DIAGNOSTIC_DESCRIPTIONS[dimension],
            "n": int(n),
            "positive_responses": int(vals.sum()),
            "observed_rate": float(vals.mean()),
        })

    return rows
````

## Transparent Bayesian provenance benchmark

The Bayesian benchmark knows only the five prototype provenance families, not GPT-5.2's actual hidden vector. It uses the synthetic diagnostic packet to update a posterior over the five hypotheses.

This provides a transparent benchmark for:

- posterior entropy;
- Bayes-optimal action by stage;
- EVPI by stage;
- expected incremental value of deeper provenance.


````python
def binom_log_pmf(k, n, p):
    p = min(1 - 1e-9, max(1e-9, float(p)))

    return (
        math.lgamma(n + 1)
        - math.lgamma(k + 1)
        - math.lgamma(n - k + 1)
        + k * math.log(p)
        + (n - k) * math.log(1 - p)
    )

def posterior_from_packet(packet, prior=None):
    if prior is None:
        prior = {
            h: 1.0 / len(HYPOTHESES)
            for h in HYPOTHESES
        }

    logw = {}

    for h in HYPOTHESES:
        ll = math.log(max(1e-15, prior[h]))

        for row in packet:
            d = row["dimension"]
            n = int(row["n"])
            k = int(row["positive_responses"])
            p = HYPOTHESIS_CENTERS[h][d]

            ll += binom_log_pmf(k, n, p)

        logw[h] = ll

    m = max(logw.values())

    weights = {
        h: math.exp(logw[h] - m)
        for h in HYPOTHESES
    }

    z = sum(weights.values())

    return {
        h: weights[h] / z
        for h in HYPOTHESES
    }

def posterior_entropy(posterior):
    p = np.array(
        [posterior[h] for h in HYPOTHESES],
        dtype=float,
    )
    p = p[p > 0]
    return float(-(p * np.log2(p)).sum())

def prototype_stage_loss(
    hypothesis,
    stage,
    action,
    severity,
    t_harm,
    information_delay=0.0,
    information_cost=0.0,
):
    return stage_loss(
        HYPOTHESIS_CENTERS[hypothesis],
        severity,
        t_harm,
        stage,
        action,
        information_delay=information_delay,
        information_cost=information_cost,
    )["total_loss"]

def best_action_under_posterior(
    stage,
    posterior,
    severity,
    t_harm,
    information_delay=0.0,
    information_cost=0.0,
):
    expected = {}

    for action in ACTIONS_BY_STAGE[stage]:
        expected[action] = sum(
            posterior[h]
            * prototype_stage_loss(
                h,
                stage,
                action,
                severity,
                t_harm,
                information_delay=information_delay,
                information_cost=information_cost,
            )
            for h in HYPOTHESES
        )

    best = min(expected, key=expected.get)

    return best, expected

def evpi_by_stage(
    stage,
    posterior,
    severity,
    t_harm,
):
    current_action, current_expected = (
        best_action_under_posterior(
            stage,
            posterior,
            severity,
            t_harm,
        )
    )

    current_risk = current_expected[current_action]

    perfect_info_risk = 0.0

    for h in HYPOTHESES:
        best_h = min(
            prototype_stage_loss(
                h,
                stage,
                action,
                severity,
                t_harm,
            )
            for action in ACTIONS_BY_STAGE[stage]
        )

        perfect_info_risk += posterior[h] * best_h

    return max(
        0.0,
        float(current_risk - perfect_info_risk),
    )
````

## Monte-Carlo expected value of deeper provenance

After the fast packet, the benchmark estimates the value of waiting for the additional observations that would complete the deep packet.

For stage $s$,

$
\operatorname{NetEVSI}_{s}
=
R_s(b_{\mathrm{fast}})
-
\mathbb E
\left[
R_s(b_{\mathrm{deep}})
\right]
-
C_{\mathrm{deep}}
-
C_{D,s}.
$

The same deeper evidence can therefore be positive-value at remediation and negative-value at containment.


````python
def packet_to_counts(packet):
    return {
        row["dimension"]: {
            "n": int(row["n"]),
            "k": int(row["positive_responses"]),
        }
        for row in packet
    }

def counts_to_packet(counts):
    rows = []

    for d in THETA_KEYS:
        n = counts[d]["n"]
        k = counts[d]["k"]

        rows.append({
            "dimension": d,
            "description": DIAGNOSTIC_DESCRIPTIONS[d],
            "n": int(n),
            "positive_responses": int(k),
            "observed_rate": float(k / n),
        })

    return rows

def simulated_incremental_deep_evsi(
    stage,
    fast_packet,
    severity,
    t_harm,
    reps=EVSI_MONTE_CARLO_REPS,
    seed=SEED,
):
    fast_posterior = posterior_from_packet(fast_packet)

    current_action, current_expected = (
        best_action_under_posterior(
            stage,
            fast_posterior,
            severity,
            t_harm,
            information_delay=FAST_INFO_DELAY,
            information_cost=FAST_INFO_COST,
        )
    )

    current_risk = current_expected[current_action]

    fast_counts = packet_to_counts(fast_packet)
    extra_n = (
        DEEP_PROVENANCE_TRIALS
        - FAST_PROVENANCE_TRIALS
    )

    local = np.random.default_rng(
        stable_seed(seed, stage, "evsi")
    )

    future_risks = []

    hypotheses = np.array(HYPOTHESES)
    probs = np.array([
        fast_posterior[h]
        for h in HYPOTHESES
    ])

    for _ in range(reps):
        sampled_h = str(
            local.choice(hypotheses, p=probs)
        )

        combined = copy.deepcopy(fast_counts)

        for d in THETA_KEYS:
            p = HYPOTHESIS_CENTERS[sampled_h][d]
            extra_k = int(
                (local.random(extra_n) < p).sum()
            )

            combined[d]["n"] += extra_n
            combined[d]["k"] += extra_k

        deep_packet_sim = counts_to_packet(combined)
        post = posterior_from_packet(deep_packet_sim)

        _, expected = best_action_under_posterior(
            stage,
            post,
            severity,
            t_harm,
            information_delay=(
                FAST_INFO_DELAY
                + DEEP_INCREMENTAL_INFO_DELAY
            ),
            information_cost=(
                FAST_INFO_COST
                + DEEP_INCREMENTAL_INFO_COST
            ),
        )

        future_risks.append(min(expected.values()))

    expected_future_risk = float(
        np.mean(future_risks)
    )

    return float(
        current_risk
        - expected_future_risk
    )
````

## Decision-equivalence classes

Two provenance hypotheses can remain scientifically distinct but institutionally equivalent at a particular stage.

For stage $s$,

$
h_i \sim_s h_j
\quad\Longleftrightarrow\quad
a_s^*(h_i)=a_s^*(h_j).
$

This is a crucial governance distinction: an institution need not spend additional time resolving uncertainty that does not change the decision.


````python
def prototype_best_action_by_stage(
    severity=0.78,
    t_harm=3.5,
):
    rows = []

    for h in HYPOTHESES:
        row = {"hypothesis": h}

        for stage in STAGES:
            losses = {
                a: prototype_stage_loss(
                    h,
                    stage,
                    a,
                    severity,
                    t_harm,
                )
                for a in ACTIONS_BY_STAGE[stage]
            }

            row[stage] = min(
                losses,
                key=losses.get,
            )

        rows.append(row)

    return pd.DataFrame(rows)

prototype_actions = prototype_best_action_by_stage()
display(prototype_actions)
````

## GPT-4o-mini institutional policy outputs

The provenance-blind policy returns a probability distribution over actions at each stage.

The provenance-guided policy additionally returns:

- a full posterior-like probability distribution over provenance hypotheses;
- stage-specific action distributions;
- the estimated probability that **deeper provenance would change the preferred action** at that stage.

That last quantity allows a direct comparison between the model's qualitative value-of-information judgment and the independent Python EVSI benchmark.


````python
def probability_object(keys):
    return {
        "type": "object",
        "properties": {
            k: {"type": "number"}
            for k in keys
        },
        "required": list(keys),
        "additionalProperties": False,
    }

def blind_stage_schema(stage):
    return {
        "type": "object",
        "properties": {
            "action_probabilities": probability_object(
                ACTIONS_BY_STAGE[stage]
            ),
            "decision_summary": {"type": "string"},
            "main_downside": {"type": "string"},
        },
        "required": [
            "action_probabilities",
            "decision_summary",
            "main_downside",
        ],
        "additionalProperties": False,
    }

BLIND_POLICY_SCHEMA = {
    "type": "object",
    "properties": {
        "incident_summary": {"type": "string"},
        "stage_decisions": {
            "type": "object",
            "properties": {
                stage: blind_stage_schema(stage)
                for stage in STAGES
            },
            "required": STAGES,
            "additionalProperties": False,
        },
    },
    "required": [
        "incident_summary",
        "stage_decisions",
    ],
    "additionalProperties": False,
}

def guided_stage_schema(stage):
    return {
        "type": "object",
        "properties": {
            "action_probabilities": probability_object(
                ACTIONS_BY_STAGE[stage]
            ),
            "probability_deeper_provenance_changes_action": {
                "type": "number"
            },
            "decision_summary": {"type": "string"},
            "remaining_uncertainty": {"type": "string"},
        },
        "required": [
            "action_probabilities",
            "probability_deeper_provenance_changes_action",
            "decision_summary",
            "remaining_uncertainty",
        ],
        "additionalProperties": False,
    }

GUIDED_POLICY_SCHEMA = {
    "type": "object",
    "properties": {
        "provenance_probabilities": probability_object(
            ALL_HYPOTHESIS_CLASSES
        ),
        "provenance_summary": {"type": "string"},
        "stage_decisions": {
            "type": "object",
            "properties": {
                stage: guided_stage_schema(stage)
                for stage in STAGES
            },
            "required": STAGES,
            "additionalProperties": False,
        },
        "global_uncertainty": {"type": "string"},
    },
    "required": [
        "provenance_probabilities",
        "provenance_summary",
        "stage_decisions",
        "global_uncertainty",
    ],
    "additionalProperties": False,
}
````

## Structured-output schema preflight

The schemas deliberately avoid brittle array-length and numeric-bound constraints at the provider level. Those constraints are enforced locally.


````python
FORBIDDEN_WIRE_SCHEMA_KEYS = {
    "maxItems",
    "minItems",
    "minimum",
    "maximum",
    "exclusiveMinimum",
    "exclusiveMaximum",
    "minLength",
    "maxLength",
}

def find_forbidden_schema_keys(obj, path="root"):
    hits = []

    if isinstance(obj, dict):
        for k, v in obj.items():
            if k in FORBIDDEN_WIRE_SCHEMA_KEYS:
                hits.append(f"{path}.{k}")

            hits.extend(
                find_forbidden_schema_keys(
                    v,
                    f"{path}.{k}",
                )
            )

    elif isinstance(obj, list):
        for i, v in enumerate(obj):
            hits.extend(
                find_forbidden_schema_keys(
                    v,
                    f"{path}[{i}]",
                )
            )

    return hits

for name, schema in {
    "hidden_incident": HIDDEN_INCIDENT_SCHEMA,
    "blind_policy": BLIND_POLICY_SCHEMA,
    "guided_policy": GUIDED_POLICY_SCHEMA,
}.items():
    hits = find_forbidden_schema_keys(schema)

    if hits:
        raise RuntimeError(
            f"{name} contains fragile schema constraints: {hits}"
        )

    assert schema["type"] == "object"
    assert schema["additionalProperties"] is False

    print(f"{name}: schema preflight PASS")
````

## Condition A — Provenance-blind institutional policy

GPT-4o mini receives the incident and action menus but no provenance hypothesis library and no diagnostic packet.


````python
BLIND_POLICY_SYSTEM = '''
You are the institutional decision model in a safe synthetic incident-response experiment.

You receive a public incident, urgency information, and stage-specific action menus.
You do NOT receive provenance evidence or a provenance hypothesis library.

For prevention, containment, remediation, and redeployment, return a complete probability distribution across that stage's actions.
Balance residual risk, operational restriction, control cost, and time pressure.
Do not invent hidden evidence.
Do not provide a private chain-of-thought transcript; provide only concise decision summaries.
'''

def blind_policy_prompt(hidden):
    return f'''PUBLIC INCIDENT
---------------
{render_public_case(hidden)}

INSTITUTIONAL ACTION MENUS
--------------------------
{action_menu_text()}

Choose a complete stage-by-stage policy under provenance uncertainty.
'''

def normalize_blind_policy(raw):
    for stage in STAGES:
        probs = raw["stage_decisions"][stage][
            "action_probabilities"
        ]

        raw["stage_decisions"][stage][
            "action_probabilities"
        ] = normalize_distribution(
            probs,
            ACTIONS_BY_STAGE[stage],
        )

        raw["stage_decisions"][stage][
            "selected_action"
        ] = max(
            raw["stage_decisions"][stage][
                "action_probabilities"
            ],
            key=raw["stage_decisions"][stage][
                "action_probabilities"
            ].get,
        )

    return raw

def call_blind_policy(hidden):
    last_error = None

    for attempt in range(
        1,
        MAX_MODEL_RETRIES + 1,
    ):
        try:
            response = openai_client.responses.create(
                model=DECISION_MODEL,
                instructions=BLIND_POLICY_SYSTEM,
                input=blind_policy_prompt(hidden),
                text={
                    "format": {
                        "type": "json_schema",
                        "name": "provenance_blind_policy",
                        "schema": BLIND_POLICY_SCHEMA,
                        "strict": True,
                    }
                },
                max_output_tokens=DECISION_MAX_OUTPUT_TOKENS,
            )

            raw = completed_structured_response(
                response,
                "GPT-4o-mini provenance-blind policy",
            )

            return normalize_blind_policy(raw)

        except Exception as exc:
            last_error = exc
            print(
                f"  blind policy attempt {attempt} failed: "
                f"{type(exc).__name__}: {exc}"
            )

    raise RuntimeError(
        f"Blind policy failed. Last error={last_error}"
    )
````

## Conditions B and C — Fast and deep provenance

The fast condition receives a small diagnostic packet. The deep condition receives a larger packet from the same evidence streams.

GPT-4o mini must infer provenance, choose actions, and estimate whether deeper provenance would change the action at each stage.


````python
GUIDED_POLICY_SYSTEM = '''
You are the institutional provenance and incident-response decision model in a safe synthetic experiment.

You receive:
- a public incident;
- a candidate provenance library;
- noisy provenance evidence;
- stage-specific action menus;
- the information delay already consumed.

Return:
1. a complete probability distribution over provenance classes;
2. a complete action distribution for each institutional stage;
3. for each stage, the probability that materially deeper provenance would change the preferred action.

Do not assume the diagnostic evidence is perfect.
Treat containment as time-sensitive.
Treat prevention, remediation, and redeployment as decisions whose value of provenance may differ.
Do not provide a private chain-of-thought transcript.
'''

def guided_policy_prompt(
    hidden,
    packet,
    information_delay,
    evidence_label,
):
    return f'''PUBLIC INCIDENT
---------------
{render_public_case(hidden)}

PROVENANCE HYPOTHESIS LIBRARY
-----------------------------
{hypothesis_library_text()}

PROVENANCE EVIDENCE — {evidence_label}
--------------------------------------
{json.dumps(packet, indent=2)}

INFORMATION DELAY ALREADY CONSUMED
----------------------------------
{information_delay:.2f} synthetic time units

INSTITUTIONAL ACTION MENUS
--------------------------
{action_menu_text()}

Infer provenance and choose a complete stage-by-stage policy.
'''

def normalize_guided_policy(raw):
    raw["provenance_probabilities"] = (
        normalize_distribution(
            raw["provenance_probabilities"],
            ALL_HYPOTHESIS_CLASSES,
        )
    )

    raw["top_provenance"] = max(
        raw["provenance_probabilities"],
        key=raw["provenance_probabilities"].get,
    )

    for stage in STAGES:
        stage_obj = raw["stage_decisions"][stage]

        stage_obj["action_probabilities"] = (
            normalize_distribution(
                stage_obj["action_probabilities"],
                ACTIONS_BY_STAGE[stage],
            )
        )

        stage_obj["selected_action"] = max(
            stage_obj["action_probabilities"],
            key=stage_obj[
                "action_probabilities"
            ].get,
        )

        stage_obj[
            "probability_deeper_provenance_changes_action"
        ] = clip01(
            stage_obj[
                "probability_deeper_provenance_changes_action"
            ]
        )

    return raw

def call_guided_policy(
    hidden,
    packet,
    information_delay,
    evidence_label,
):
    last_error = None

    for attempt in range(
        1,
        MAX_MODEL_RETRIES + 1,
    ):
        try:
            response = openai_client.responses.create(
                model=DECISION_MODEL,
                instructions=GUIDED_POLICY_SYSTEM,
                input=guided_policy_prompt(
                    hidden,
                    packet,
                    information_delay,
                    evidence_label,
                ),
                text={
                    "format": {
                        "type": "json_schema",
                        "name": "provenance_guided_policy",
                        "schema": GUIDED_POLICY_SCHEMA,
                        "strict": True,
                    }
                },
                max_output_tokens=DECISION_MAX_OUTPUT_TOKENS,
            )

            raw = completed_structured_response(
                response,
                f"GPT-4o-mini {evidence_label} provenance policy",
            )

            return normalize_guided_policy(raw)

        except Exception as exc:
            last_error = exc
            print(
                f"  guided policy attempt {attempt} failed: "
                f"{type(exc).__name__}: {exc}"
            )

    raise RuntimeError(
        f"Guided policy failed. Last error={last_error}"
    )
````

## Independent policy scoring

GPT-4o mini never receives the private stage-loss surface.

Python evaluates every selected action against the same hidden incident. This keeps the decision model from grading itself.


````python
def score_policy(
    hidden,
    policy,
    information_delay,
    information_cost,
):
    stage_rows = []

    for stage in STAGES:
        action = policy["stage_decisions"][stage][
            "selected_action"
        ]

        score = stage_loss(
            hidden["theta"],
            hidden["severity"],
            hidden["time_to_irreversible_harm"],
            stage,
            action,
            information_delay=information_delay,
            information_cost=0.0,
        )

        stage_rows.append({
            "stage": stage,
            "selected_action": action,
            **score,
        })

    total = (
        sum(r["total_loss"] for r in stage_rows)
        + float(information_cost)
    )

    return {
        "total_lifecycle_loss": float(total),
        "information_cost_once": float(information_cost),
        "stage_scores": stage_rows,
    }

def private_oracle_policy(hidden):
    stage_decisions = {}

    for stage in STAGES:
        losses = {
            action: stage_loss(
                hidden["theta"],
                hidden["severity"],
                hidden["time_to_irreversible_harm"],
                stage,
                action,
            )["total_loss"]
            for action in ACTIONS_BY_STAGE[stage]
        }

        best = min(losses, key=losses.get)

        stage_decisions[stage] = {
            "selected_action": best,
            "private_losses": losses,
        }

    return {
        "stage_decisions": stage_decisions,
    }

def score_private_oracle(hidden):
    policy = private_oracle_policy(hidden)

    rows = []

    for stage in STAGES:
        action = policy["stage_decisions"][stage][
            "selected_action"
        ]

        rows.append({
            "stage": stage,
            "selected_action": action,
            **stage_loss(
                hidden["theta"],
                hidden["severity"],
                hidden["time_to_irreversible_harm"],
                stage,
                action,
            ),
        })

    return {
        "total_lifecycle_loss": float(
            sum(r["total_loss"] for r in rows)
        ),
        "stage_scores": rows,
    }
````

## Run the complete experiment

This is the expensive cell.

For every hidden incident:

1. GPT-5.2 constructs the private provenance case.
2. GPT-4o mini chooses a provenance-blind policy.
3. GPT-4o mini chooses a fast-provenance policy.
4. GPT-4o mini chooses a deep-provenance policy.
5. Python scores all policies.
6. Python computes Bayesian stage-specific VOI benchmarks.


````python
hidden_incidents = []
policy_records = []
voi_records = []

case_index = 0

for hypothesis in HYPOTHESES:
    for rep in range(N_REPS_PER_HYPOTHESIS):
        case_index += 1
        case_seed = (
            SEED
            + 1000 * case_index
            + rep
        )

        case_id = f"{hypothesis}_{rep+1}"

        print(
            f"\n=== {case_id} | seed {case_seed} ==="
        )

        hidden = build_hidden_incident(
            hypothesis,
            case_seed,
        )

        hidden["case_id"] = case_id
        hidden_incidents.append(hidden)

        print(
            f"hidden incident accepted | "
            f"decoy={hidden['decoy']} | "
            f"severity={hidden['severity']:.2f} | "
            f"T_H={hidden['time_to_irreversible_harm']:.2f}"
        )

        fast_packet = diagnostic_packet(
            hidden,
            FAST_PROVENANCE_TRIALS,
        )

        deep_packet = diagnostic_packet(
            hidden,
            DEEP_PROVENANCE_TRIALS,
        )

        blind = call_blind_policy(hidden)
        fast = call_guided_policy(
            hidden,
            fast_packet,
            FAST_INFO_DELAY,
            "FAST",
        )
        deep = call_guided_policy(
            hidden,
            deep_packet,
            (
                FAST_INFO_DELAY
                + DEEP_INCREMENTAL_INFO_DELAY
            ),
            "DEEP",
        )

        blind_score = score_policy(
            hidden,
            blind,
            information_delay=0.0,
            information_cost=0.0,
        )

        fast_score = score_policy(
            hidden,
            fast,
            information_delay=FAST_INFO_DELAY,
            information_cost=FAST_INFO_COST,
        )

        deep_score = score_policy(
            hidden,
            deep,
            information_delay=(
                FAST_INFO_DELAY
                + DEEP_INCREMENTAL_INFO_DELAY
            ),
            information_cost=(
                FAST_INFO_COST
                + DEEP_INCREMENTAL_INFO_COST
            ),
        )

        oracle_score = score_private_oracle(hidden)

        for condition, policy, score in [
            ("blind", blind, blind_score),
            ("fast", fast, fast_score),
            ("deep", deep, deep_score),
        ]:
            policy_records.append({
                "case_id": case_id,
                "hypothesis": hypothesis,
                "condition": condition,
                "policy": policy,
                "score": score,
                "fast_packet": (
                    fast_packet
                    if condition in {"fast", "deep"}
                    else None
                ),
                "deep_packet": (
                    deep_packet
                    if condition == "deep"
                    else None
                ),
            })

        policy_records.append({
            "case_id": case_id,
            "hypothesis": hypothesis,
            "condition": "private_oracle",
            "policy": private_oracle_policy(hidden),
            "score": oracle_score,
            "fast_packet": None,
            "deep_packet": None,
        })

        fast_posterior = posterior_from_packet(
            fast_packet
        )
        deep_posterior = posterior_from_packet(
            deep_packet
        )

        for stage in STAGES:
            voi_records.append({
                "case_id": case_id,
                "hypothesis": hypothesis,
                "stage": stage,
                "fast_entropy_bits": posterior_entropy(
                    fast_posterior
                ),
                "deep_entropy_bits": posterior_entropy(
                    deep_posterior
                ),
                "evpi_after_fast": evpi_by_stage(
                    stage,
                    fast_posterior,
                    hidden["severity"],
                    hidden["time_to_irreversible_harm"],
                ),
                "net_evsi_deep_after_fast": (
                    simulated_incremental_deep_evsi(
                        stage,
                        fast_packet,
                        hidden["severity"],
                        hidden["time_to_irreversible_harm"],
                        seed=case_seed,
                    )
                ),
                "llm_probability_deeper_changes_action": (
                    fast["stage_decisions"][stage][
                        "probability_deeper_provenance_changes_action"
                    ]
                ),
                "fast_selected_action": (
                    fast["stage_decisions"][stage][
                        "selected_action"
                    ]
                ),
                "deep_selected_action": (
                    deep["stage_decisions"][stage][
                        "selected_action"
                    ]
                ),
                "realized_action_changed": (
                    fast["stage_decisions"][stage][
                        "selected_action"
                    ]
                    !=
                    deep["stage_decisions"][stage][
                        "selected_action"
                    ]
                ),
            })

        print(
            "lifecycle loss | "
            f"blind={blind_score['total_lifecycle_loss']:.3f} | "
            f"fast={fast_score['total_lifecycle_loss']:.3f} | "
            f"deep={deep_score['total_lifecycle_loss']:.3f} | "
            f"oracle={oracle_score['total_lifecycle_loss']:.3f}"
        )

        time.sleep(SLEEP_BETWEEN_CASES)

print(
    f"\nCompleted {len(hidden_incidents)} incidents, "
    f"{len(policy_records)} policy evaluations, "
    f"and {len(voi_records)} stage-level VOI observations."
)
````

## Decision-Relevance Report

The original Notebook 10 ended with a decision-relevance report. The revised version expands that artifact.

The report now includes:

- observed effect and timing;
- provenance posterior;
- residual entropy;
- high-probability unresolved alternatives;
- decision-equivalence groups by stage;
- stage actions under fast and deep evidence;
- stage-specific EVPI and net EVSI;
- whether deeper provenance is normatively worth acquiring;
- remediation recommendation;
- redeployment condition;
- explicit scientific status.

It remains an advisory report, not a safety certificate.


````python
def decision_relevance_report(
    hidden,
    fast_policy,
    deep_policy,
    fast_packet,
):
    fast_posterior = posterior_from_packet(
        fast_packet
    )

    report = {
        "case_id": hidden["case_id"],
        "observed_effect": hidden[
            "public_case"
        ]["observed_effect"],
        "timing": {
            "T_H": hidden[
                "time_to_irreversible_harm"
            ],
            "fast_information_delay": (
                FAST_INFO_DELAY
            ),
            "deep_total_information_delay": (
                FAST_INFO_DELAY
                + DEEP_INCREMENTAL_INFO_DELAY
            ),
        },
        "fast_bayesian_posterior": (
            fast_posterior
        ),
        "fast_posterior_entropy_bits": (
            posterior_entropy(
                fast_posterior
            )
        ),
        "llm_fast_provenance_probabilities": (
            fast_policy[
                "provenance_probabilities"
            ]
        ),
        "supported_hypotheses": sorted(
            fast_posterior,
            key=fast_posterior.get,
            reverse=True,
        )[:3],
        "stage_decisions": {},
        "decision_equivalence_groups": {
            stage: decision_equivalence_groups(
                stage
            )
            for stage in STAGES
        },
        "scientific_status": (
            "conditional_on_synthetic_provenance_geometry_"
            "action_losses_model_versions_and_seed"
        ),
    }

    for stage in STAGES:
        evpi = evpi_by_stage(
            stage,
            fast_posterior,
            hidden["severity"],
            hidden[
                "time_to_irreversible_harm"
            ],
        )

        net_evsi = (
            simulated_incremental_deep_evsi(
                stage,
                fast_packet,
                hidden["severity"],
                hidden[
                    "time_to_irreversible_harm"
                ],
                seed=hidden[
                    "case_seed"
                ],
            )
        )

        report["stage_decisions"][stage] = {
            "fast_action": (
                fast_policy[
                    "stage_decisions"
                ][stage][
                    "selected_action"
                ]
            ),
            "deep_action": (
                deep_policy[
                    "stage_decisions"
                ][stage][
                    "selected_action"
                ]
            ),
            "EVPI_after_fast": evpi,
            "net_EVSI_deeper_provenance": (
                net_evsi
            ),
            "deeper_provenance_worthwhile": (
                net_evsi > 0
            ),
        }

    remediation_action = (
        deep_policy[
            "stage_decisions"
        ]["remediation"][
            "selected_action"
        ]
    )

    redeployment_action = (
        deep_policy[
            "stage_decisions"
        ]["redeployment"][
            "selected_action"
        ]
    )

    report[
        "recommended_remediation"
    ] = remediation_action

    report[
        "redeployment_condition"
    ] = (
        "restricted_or_shadow_redeployment"
        if redeployment_action
        in {"bounded", "quarantine", "shadow_mode"}
        else "full_redeployment_only_after_validation"
    )

    return report
````

````python
# Build and print one representative report after the experiment runs.
if hidden_incidents:
    example_hidden = hidden_incidents[0]

    example_fast = next(
        r for r in policy_records
        if r["case_id"]
        == example_hidden["case_id"]
        and r["condition"] == "fast"
    )

    example_deep = next(
        r for r in policy_records
        if r["case_id"]
        == example_hidden["case_id"]
        and r["condition"] == "deep"
    )

    example_report = (
        decision_relevance_report(
            example_hidden,
            example_fast["policy"],
            example_deep["policy"],
            example_fast["fast_packet"],
        )
    )

    print(
        json.dumps(
            example_report,
            indent=2,
        )
    )
````

## Interpretation guide

The experiment is intentionally designed so that deeper provenance is not automatically beneficial.

### Pattern 1 — Deep provenance sharply lowers entropy but net EVSI is negative for containment

This is a central Notebook 10 result. It means that the evidence is scientifically informative but operationally unnecessary for immediate containment. Several remaining hypotheses imply the same robust control, and waiting for attribution destroys more value than the extra precision creates.

### Pattern 2 — Deep provenance has positive EVSI for remediation

This indicates that provenance changes the root-cause repair. The institution benefits from distinguishing, for example, a common-mode problem from a reward problem or external orchestration.

### Pattern 3 — Deep provenance has high value for redeployment

This means the admissible autonomy envelope depends materially on attribution. A system that can be safely restored after a well-understood communication problem may require bounded or shadow-mode operation if strategic opacity remains unresolved.

### Pattern 4 — Attribution remains uncertain but actions are decision-equivalent

This is not a failure. If the high-posterior hypotheses imply the same optimal action, further provenance can have low institutional value. The appropriate report is not “cause unknown, therefore unsafe,” but rather “remaining causal ambiguity does not alter the current decision.”

### Pattern 5 — GPT-4o mini wants deeper provenance when benchmark EVSI is negative

This is an information-governance failure: the model is overvaluing explanation relative to action.

### Pattern 6 — GPT-4o mini rejects deeper provenance when benchmark EVSI is positive

This is the opposite failure: the model underinvests in attribution precisely where remediation or redeployment depends on it.

### Pattern 7 — Deep provenance improves diagnosis but worsens lifecycle loss

This connects directly to Notebook 9. More accurate provenance arrived too late or cost too much to improve the final institutional outcome.

The strongest result is therefore not maximal attribution accuracy. It is **stage-appropriate informational discipline**.


## Conclusion

Notebook 10 completes the methodological progression from hidden strategic structure to institutional governance of uncertainty. The original ten-notebook sequence ended with the insight that perfect attribution is not always required before protective action. The revised Notebook 10 retains that conclusion but places it inside a substantially richer decision-theoretic structure.

The first major result of the redesign is conceptual: **provenance is not intrinsically valuable**. It becomes valuable through the decisions it changes. A diagnostic packet can contain substantial information, reduce posterior entropy, and improve classification accuracy while producing almost no institutional benefit at a stage where all plausible hypotheses imply the same action. Conversely, a comparatively modest reduction in uncertainty can be highly valuable if it separates hypotheses that imply different remediation architectures or different redeployment boundaries.

This distinction is captured formally through stage-specific Bayes risk, EVPI, and EVSI. For each institutional stage, the relevant object is not simply the posterior over provenance hypotheses but the loss surface induced by that posterior:

$
R_s(b)
=
\min_a
\mathbb E_b[L_s(a,\theta)].
$

The expected value of perfect information identifies the maximum amount the institution should ever be willing to pay to eliminate uncertainty at that stage. The expected value of deeper provenance asks a more realistic question: how much is the next evidence package worth, given what is already known? Once information cost and delay are subtracted, the resulting net EVSI becomes a direct measure of whether additional attribution work is institutionally justified.

The second major result is the separation between **scientific equivalence and decision equivalence**. Notebook 7 established that representation can make strategically equivalent systems look different to a bounded observer. Notebook 8 showed that active experimentation may be required to recover the hidden mechanism. Notebook 10 adds the observation that two scientifically distinct provenance hypotheses can nevertheless be decision-equivalent at a particular stage. If both imply `pause_actions` during containment, the institution may not need to distinguish them before acting. If they imply different remediation actions later, the same distinction becomes important after the system is stabilized.

This yields a useful governance rule:

$
\boxed{
\text{resolve uncertainty only to the level required by the next consequential decision.}
}
$

That rule is more disciplined than either extreme. It rejects the idea that action must wait for complete causal attribution, but it also rejects the idea that provenance is merely an academic luxury. The proper level of provenance is endogenous to the stage, the available action set, the time to harm, the control cost, and the consequences of choosing incorrectly.

The third major result is the institutional importance of **strategic opacity as provenance**. The `opaque_familiar_game` hypothesis carries forward the core finding of Notebook 7: difficult recognition does not imply an unfamiliar underlying strategic structure. A transformed representation can conceal a familiar game from a bounded observer. For Notebook 10, this matters because remediation may require increasing legibility or reconstructing and validating the strategic mechanism rather than disabling communication, redesigning rewards, or removing an external controller. Provenance therefore includes not only “who caused the behavior?” but also “what strategic structure generated it, and can that structure be reconstructed reliably?”

The fourth major result concerns **timing**. Notebook 9 established that diagnostic value can change sign when intervention delay is counted. Notebook 10 makes that timing cost stage-specific. Containment is the most obvious case: if a broad action is already robust across high-probability provenance hypotheses, deep attribution can be negative-value because the institutional decision is already clear and time-sensitive. Prevention and remediation can often tolerate more analysis. Redeployment may place high value on provenance because the institution is deciding whether to restore permissions, autonomy, and operational scope.

The resulting lifecycle logic is not simply:

$
\text{diagnose}
\rightarrow
\text{act}.
$

It is better represented as:

$
\boxed{
\text{contain at the resolution required now}
\rightarrow
\text{preserve evidence}
\rightarrow
\text{acquire deeper provenance when decision-relevant}
\rightarrow
\text{remediate}
\rightarrow
\text{validate}
\rightarrow
\text{redeploy only within the supported autonomy envelope}.
}
$

This logic turns provenance into institutional capital. Evidence preservation creates an option to improve later decisions. Destroying or failing to record provenance can therefore have an economic cost even when it does not affect the immediate containment outcome. In a real institution, this idea connects provenance to logging architecture, identity systems, model registries, immutable audit records, decision traces, tool-use records, and reproducible incident reconstruction. The notebook does not implement those systems, but it provides the decision-theoretic reason they can be valuable.

The fifth major result is methodological. GPT-5.2 does not decide the outcome of the experiment. It constructs semantically rich hidden incidents inside a private causal grammar. Python independently constrains the mechanism, generates evidence, evaluates stage losses, and computes value-of-information benchmarks. GPT-4o mini is a bounded institutional decision model evaluated under different information regimes. This separation of case generation, inference, and scoring is essential if the notebook is to study reasoning rather than prompt compliance.

The sixth major result is that provenance quality should not be evaluated solely through classification metrics. A model can misclassify the provenance family yet choose an institutionally robust action and incur low regret. A model can also identify the family correctly yet overreact, underreact, or spend too much time acquiring information that does not change the decision. The revised notebook therefore evaluates lifecycle loss, stage loss, regret, action changes, posterior entropy, EVPI, EVSI, and the model's own assessment of whether deeper provenance is likely to alter the action.

This creates a direct bridge to information economics. Provenance acquisition resembles any other costly information technology. Its value depends on the decision problem. In a perfectly symmetric decision environment, additional information has little value because all states lead to the same action. In a highly state-contingent environment, information can be extremely valuable because it changes the optimal policy. The cybersecurity and AI-governance significance is that incident-response architectures should be designed around **decision-relevant observability**, not indiscriminate data accumulation.

The notebook also identifies an important stopping problem. At any stage, the institution can ask:

$
\text{Do we know enough to act?}
$

Formally, deeper provenance should be acquired only while its expected reduction in decision loss exceeds its cost and delay. This can be interpreted as an institutional stopping boundary between investigation and action. In future work, that boundary can be embedded in a fully dynamic POMDP or continuous-time stochastic control problem in which evidence arrives endogenously, incidents evolve while being investigated, and actions change the future information structure.

Several limitations remain. The provenance families are abstractions. The latent causal dimensions are stylized. The diagnostic observations are Bernoulli measurements rather than real telemetry. The action costs are synthetic. GPT-4o mini sees concise evidence summaries rather than raw security logs. Real-world provenance also depends on data integrity, cryptographic attestation, identity assurance, model versioning, chain-of-custody, and organizational process. None of those implementation details are estimated here.

Those limitations do not undermine the central proposition. They define the boundary of what the experiment can claim.

The strongest permissible conclusion is:

$
\boxed{
\text{Provenance has institutional value when—and only when—}
\\
\text{it changes a consequential decision enough to justify the cost and delay of acquiring it.}
}
$

The institutional objective is therefore not perfect attribution at every moment. It is **decision-relevant provenance at the right stage, at the right resolution, before the value of the decision window disappears**.

That completes the progression from Notebook 7 through Notebook 10:

$
\boxed{
\text{representation}
\rightarrow
\text{diagnosis}
\rightarrow
\text{intervention}
\rightarrow
\text{provenance-aware institutional governance}.
}
$
